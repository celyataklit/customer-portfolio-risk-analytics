## 2. Import Libraries

## 3. Initial Data Quality Checks
This section evaluates dataset size, missing values, and overall data quality before transformation and cleaning.

```python
df.isna().sum()
```

## 4. Data Type Conversion

```python
# convert numeric columns

df["CustAccountBalance"] = pd.to_numeric(
    df["CustAccountBalance"],
    errors= "coerce"
)

df["TransactionAmount (INR)"] = pd.to_numeric(
    df["TransactionAmount (INR)"],
    errors="coerce"
)

# Convert dates
df["TransactionDate"] = pd.to_datetime(
    df["TransactionDate"],
    format="%d/%m/%y",
    errors="coerce"
)

df["CustomerDOB"] = pd.to_datetime(
    df["CustomerDOB"],
    format="%d/%m/%y",
    errors="coerce"
)
```

## 5. Missing Values Treatment & SQL Compatibility

```python

# Remove missing values
df_clean = df.dropna()

# Rename column
df_clean = df_clean.rename(columns={
    "TransactionAmount (INR)": "TransactionAmount"
})

# Final check
print(df_clean.shape)

print(df_clean.isna().sum())
```

## 6. Export Clean Dataset

```python
df_clean.to_csv(
    r"C:\Users\LENOVO\Desktop\Projet 3\bank_transactions_clean.csv",
    index=False
)

print("Clean dataset exported successfully")
```
